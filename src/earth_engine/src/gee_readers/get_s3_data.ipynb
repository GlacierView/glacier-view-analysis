```python
import boto3
import pandas as pd
import os
from tqdm import tqdm
```

```python
bucket_name = 'full-time-series-images-t1-l2-sr'
```

```python
# Create a Boto3 S3 client
s3_client = boto3.client('s3')

# Retrieve all objects in the specified bucket
objects = s3_client.list_objects_v2(Bucket=bucket_name, Delimiter='/')
s3_glims_ids = [list(val.values())[0][:-1] for val in objects['CommonPrefixes']]
```

```python
s3 = boto3.client('s3')
s3_glims_id = s3_glims_ids[0]
paginator = s3.get_paginator('list_objects_v2')
pages = paginator.paginate(Bucket=bucket_name, Prefix=s3_glims_id)

for page in pages:
    for obj in tqdm(page['Contents']):
        file_name = f"/Users/mattw/Desktop/projects/GlacierView/src/earth_engine/data/ee_landing_zone/full_time_series_c02_t1_l2/{obj['Key']}"
        print(file_name)
        if file_name.endswith(".DS_Store"): continue
        s3.download_file(
            bucket_name,
            obj['Key'],
            file_name)

```
