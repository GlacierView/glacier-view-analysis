```python
pulled_files = []
for glims_id in os.listdir(data_dir):
    glims_id_dir = os.path.join(data_dir, glims_id)
    try:
        for file_name in os.listdir(glims_id_dir):
           pulled_files.append((glims_id, file_name))
    except NotADirectoryError:
        continue
```

```python
pulled_df = pd.DataFrame(pulled_files, columns = ['glims_id', 'file_name'])
```
