The purpose of this notebook is to pull DEMs from earth engine. Additionally, logs are created with each image pulled from earth engine. Monitoring the logs as the imagery is being pulled is a helpful way to ensure images are still being pulled as the logs are updated in real time.

```python
import os

import ee
import ee_helpers
import geopandas as gpd
import pandas as pd

from glacierview import config, preprocess
from glacierview import rasters as read  # was helpers/read.py
from glacierview import viz as explore  # was helpers/explore.py
from glacierview.earthengine import EePull
```

```python
# Trigger the authentication flow
ee.Authenticate()
# Initialize the library
ee.Initialize()
```

```python
data_label = "full_time_series_c02_t1_l2"
glacier_view_dir = str(config.REPO_ROOT)
glims_dir = os.path.join(glacier_view_dir, "src", "glims")
log_dir = os.path.join(glacier_view_dir,"src","earth_engine","data","ee_landing_zone",data_label, "logs")
out_dir = os.path.join(glacier_view_dir,"src","earth_engine","data","ee_landing_zone",data_label, "dems")
processed_metadata_output_dir = os.path.join(glacier_view_dir, "src", "earth_engine", "data", "processed_metadata", data_label)
```

```python
ee_pull = EePull(log_dir, "geog_area_rollup_250")
```

```python
df_sample = pd.read_csv(os.path.join(glims_dir, "data", "inference_samples", "geog_area_rollup_250.csv"))
glims_ids = df_sample.glims_id
```

```python
# net_new_df = pd.read_csv(os.path.join(processed_metadata_output_dir,"net_new_glims_ids_for_segmentation.csv"))
# net_new_glims_ids = list(pd.Series(glims_ids)[pd.Series(glims_ids).isin(net_new_df.glims_id)])
```

```python
glims_bb_path = os.path.join(glims_dir,"data", "training_samples","glims_18k_bb.shp")
df = gpd.read_file(glims_bb_path)
```

```python
df = df[df.glac_id.isin(glims_ids)]
```

```python
# if len(os.listdir(out_dir)) > 0:
#     glis_id_dedupe = [file_name.split("_")[0] for file_name in os.listdir(out_dir)]
#     df = df[~df.glac_id.isin(glims_id_dedupe)]
```

```python
for row in df.iterrows():
    glims_id = row[1].glac_id
    bounding_box = eval(row[1].bboxes)
    ee_pull.export_nasa_dems(glims_id, bounding_box, out_dir)
```
